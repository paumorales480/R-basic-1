```r
url_zip <- "https://www.inegi.org.mx/contenidos/programas/enut/2024/microdatos/enut_2024_bd_csv.zip"
```

```r
# Crear un directorio temporal para guardar el ZIP
temp_dir <- tempdir()
zip_path <- file.path(temp_dir, "archivo.zip")
```

```r
# Descargar el ZIP
download.file(url_zip, destfile = zip_path, mode = "wb")

# Descomprimir los archivos dentro del ZIP al mismo directorio temporal
unzip(zip_path, exdir = temp_dir)

# Ver qué archivos se extrajeron
list.files(temp_dir)
```

```r
# Localizar el CSV (ajusta el nombre según el archivo que haya dentro)
csv_file <- list.files(temp_dir, pattern = "tmodulo.csv", full.names = TRUE)[1]   #Aquí solo se modifica el nombre del archivo,
#dependiendo del que quiera :)

# Leer el CSV
datos <- read_csv(csv_file)
```

#### Variables que se rescatarán de la base "t_modulo" 

cve_ent: Clave de la entidad


sexo: sexo de la persona


p6_18: Durante la semana pasada, en su tiempo libre ¿usted hizo deporte o ejercicio físico? (fútbol, basquetbol, natación, box, karate, caminar, correr, patinar, andar en bicicleta, yoga, zumba)


p6_18a_1: ¿Cuánto tiempo le dedicó de lunes a viernes? - Horas (lunes-viernes)


p6_18a_2: ¿Cuánto tiempo le dedicó de lunes a viernes? - Minutos (lunes-viernes)


p6_18a_3: ¿Cuánto tiempo le dedicó de lunes a viernes? - Horas (sábado)


p6_18a_4: ¿Cuánto tiempo le dedicó de lunes a viernes? - Minutos (domingo)


variables diseño muestral: est_dis, upm_dis, fac_per

```r
datos <- datos %>%
  select(CVE_ENT, SEXO, P6_18, P6_18A_1, P6_18A_2, P6_18A_3, P6_18A_4, EDAD_V, EST_DIS, UPM_DIS, FAC_PER) %>%
  clean_names() %>%
  mutate(across(c(p6_18a_1, p6_18a_2, p6_18a_3, p6_18a_4), as.numeric)) %>%
  mutate(
    sexo = case_when(
      sexo == 1 ~ "Hombre",
      sexo == 2 ~ "Mujer",
      TRUE ~ "No especificado"
    ),
    deporte = case_when(
      p6_18 == 1 ~ "Sí",
      p6_18 == 2 ~ "No",
      TRUE ~ "No especificado"
    ),
    tiempo_total_semanal = (p6_18a_1 + (p6_18a_2 / 60) + p6_18a_3 + (p6_18a_4 / 60)),
    edad = case_when(
      edad_v >= 12 & edad_v <= 29 ~ "Personas jóvenes (12-29)",
      edad_v > 29 & edad_v <= 64 ~ "29-64",
      edad_v >= 65 ~ "Adultos mayores",
      TRUE ~ NA_character_
    )
  ) %>%
  filter(as.character(cve_ent) %in% c("9", "09"))

head(datos)
```

```r
# Diseño muestral 
dm <- datos %>%
  as_survey_design(
    ids = upm_dis,
    strata = est_dis,
    weights = fac_per,
    nest = TRUE
  )
```

```r
cat("Total de la población de 12 años y más en la CDMX que respondieron de forma afirmativa
que durante la semana pasada, en su tiempo libre hicieron deporte o ejercicio físico
(fútbol, basquetbol, natación, box, karate, caminar, correr, patinar,
andar en bicicleta, yoga, zumba)\n")

deporte <- dm %>%
  group_by(deporte) %>%
  summarise(
    porcentaje_poblacion = survey_prop(vartype = "cv"),
    total_poblacion_n = survey_total(vartype = "cv")) %>%
  mutate(
    porcentaje_poblacion = round(porcentaje_poblacion * 100, 2)
  )
deporte
```

```r
cat("Distribución por sexo de la población de 12 años y más en la CDMX que respondieron de forma afirmativa
que durante la semana pasada, en su tiempo libre hicieron deporte o ejercicio físico
(fútbol, basquetbol, natación, box, karate, caminar, correr, patinar,
andar en bicicleta, yoga, zumba)\n")

deporte_sexo <- dm %>%
filter(deporte == "Sí") %>%
  group_by(deporte, sexo) %>%
  summarise(
    porcentaje_poblacion = survey_prop(vartype = "cv"),
    total_poblacion_n = survey_total(vartype = "cv")) %>%
  mutate(
    porcentaje_poblacion = round(porcentaje_poblacion * 100, 2)
  )
deporte_sexo
```

```r
cat("Horas promedio a la semana que la población de 12 años y más en la CDMX dedicó en su tiempo libre a hacer deporte o ejercicio físico
(fútbol, basquetbol, natación, box, karate, caminar, correr, patinar,
andar en bicicleta, yoga, zumba)\n")
tiempo_prom_deporte <- dm %>%
    filter(deporte == "Sí") %>%
  group_by(sexo) %>%
  summarise(
    horas_promedio_semana_deporte = survey_mean(tiempo_total_semanal, vartype = "cv", na.rm = TRUE)
  )%>%
  mutate(horas_promedio_semana_deporte = round(horas_promedio_semana_deporte, 2))
tiempo_prom_deporte
```

```r
cat("Horas promedio a la semana que la población de 12 años y más en la CDMX dedicó en su tiempo libre a hacer deporte o ejercicio físico
(fútbol, basquetbol, natación, box, karate, caminar, correr, patinar,
andar en bicicleta, yoga, zumba)\n")
tiempo_prom_deporte <- dm %>%
    filter(deporte == "Sí") %>%
  group_by(sexo, edad) %>%
  summarise(
    horas_promedio_semana_deporte = survey_mean(tiempo_total_semanal, vartype = "cv", na.rm = TRUE)
  )%>%
  mutate(horas_promedio_semana_deporte = round(horas_promedio_semana_deporte, 2))
tiempo_prom_deporte
```
